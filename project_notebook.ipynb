## Question 2: Resource Allocation Module (Fractional Knapsack)
**Greedy Choice:** Sort items in descending order of value-to-weight ratio ($v_i / w_i$).
**Time Complexity:** $O(n \log n)$ | **Space Complexity:** $O(n)$

```python
from algorithms.fractional_knapsack import fractional_knapsack

items = [
    {'name': 'Gold', 'weight': 10, 'value': 60},
    {'name': 'Silver', 'weight': 20, 'value': 100},
    {'name': 'Bronze', 'weight': 30, 'value': 120},
]
capacity = 50.0

max_value, allocation = fractional_knapsack(items, capacity)

print(f'=== Fractional Knapsack Allocation (Capacity: {capacity}) ===')
print(f'Max Achievable Profit: ${max_value:.2f}')
print('Selected Allocation Breakdown:')
for name, frac, wt, val in allocation:
    print(f'  {name}: {frac*100:.1f}% taken | Weight: {wt:.1f}kg | Contributed: ${val:.2f}')
```

## Question 3: Task Scheduling Module (Activity Selection)
**Greedy Choice:** Earliest Finish Time (EFT) First.
**Time Complexity:** $O(n \log n)$ | **Space Complexity:** $O(n)$

```python
from algorithms.activity_selection import activity_selection

activities = [
    {'name': 'A1', 'start': 1, 'finish': 4},
    {'name': 'A2', 'start': 3, 'finish': 5},
    {'name': 'A3', 'start': 0, 'finish': 6},
    {'name': 'A4', 'start': 5, 'finish': 7},
    {'name': 'A5', 'start': 3, 'finish': 9},
    {'name': 'A6', 'start': 5, 'finish': 9},
    {'name': 'A7', 'start': 6, 'finish': 10},
    {'name': 'A8', 'start': 8, 'finish': 11},
]

selected = activity_selection(activities)
print(f'=== Task Scheduling: Selected {len(selected)} Mutually Compatible Activities ===')
for idx, act in enumerate(selected, 1):
    print(f"  {idx}. {act['name']}: [{act['start']} -> {act['finish']}] (Duration: {act['duration']}h)")
```

## Question 4: Data Compression Module (Huffman Coding)
**Greedy Choice:** Min-Priority Queue merging of lowest frequency subtrees.
**Time Complexity:** $O(n \log n)$ | **Space Complexity:** $O(n)$

```python
from algorithms.huffman_coding import huffman_encode, huffman_decode, build_huffman_tree

sample_text = 'abracadabra'
encoded_bits, codebook, freqs, stats = huffman_encode(sample_text)

tree_root = build_huffman_tree(sample_text)
decoded_text = huffman_decode(encoded_bits, tree_root)

print('=== Huffman Coding Compression Results ===')
print(f"Original Text: '{sample_text}'")
print(f"Huffman Codebook: {codebook}")
print(f"Encoded Bits: {encoded_bits}")
print(f"Original ASCII Bits: {stats['original_bits']} bits")
print(f"Compressed Bits: {stats['compressed_bits']} bits")
print(f"Compression Ratio: {stats['compression_ratio']:.2%}")
print(f"Space Savings: {stats['space_savings_percent']:.2f}%")
print(f"Lossless Verification: Decompressed matches original exactly: {decoded_text == sample_text}")
```

## Question 7: Analytical Report & Key Observations
---
### 1. Greedy Choice & Matroid Optimality
All three algorithms satisfy both the **greedy-choice property** and **optimal substructure**:
* **Fractional Knapsack:** Proved optimal via the *Exchange Argument*. Substituting higher ratio weight always improves total value.
* **Activity Selection:** Proved optimal because finishing earliest leaves maximum remaining time for subsequent activities.
* **Huffman Coding:** Proved optimal via Shannon-Fano entropy bounds and min-heap prefix property.

### 2. When Does Greedy Fail? (0/1 Knapsack vs Fractional Knapsack)
When items cannot be fractionally divided, the greedy choice of picking highest ratio items fails. An item with high ratio may occupy space that prevents taking two items with higher combined value. 0/1 Knapsack requires **Dynamic Programming** ($O(nW)$ pseudo-polynomial time).

### 3. Practical Applications
* **Fractional Knapsack:** Cloud compute slicing (AWS/GCP), financial portfolio fractional asset investment.
* **Activity Selection:** Single-resource processor scheduling, conference room and lecture hall allocation.
* **Huffman Coding:** Core entropy compression in JPEG, MP3, GZIP, and ZIP formats.
